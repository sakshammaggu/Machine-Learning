```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
df
```

```python
titanic = pd.read_csv('../data/train.csv')
titanic
```

```python
flights = sns.load_dataset('flights')
flights
```

```python
iris = sns.load_dataset('iris')
iris
```

```python
# Scatterplot (Numerical-Numerical)
sns.scatterplot(data=df, x='total_bill', y='tip', hue='smoker')
```

```python
# Barplot (Numerical-Categorical)
sns.barplot(data=titanic, x='Pclass', y='Age', hue='Sex')
```

```python
# Boxplot (Numerical-Categorical)
sns.boxplot(data=flights, x='month', y='passengers')
```

```python
# Distplot (Numerical-Categorical)
sns.displot(data=iris, x='sepal_length', hue='species', kde=True)
```

```python
# Heatmap (Categorical-Categorical)
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']), annot=True, fmt='d')
```

```python
# Clustermap (Categorical-Categorical)
sns.clustermap(pd.crosstab(iris['species'], iris['sepal_length']), annot=True, fmt='d')
```

```python
# Lineplot (Numerical-Numerical)
sns.lineplot(data=flights, x='year', y='passengers', hue='month')
```

```python
# Pairplot (Categorical-Categorical)
sns.pairplot(iris, hue='species')
```
