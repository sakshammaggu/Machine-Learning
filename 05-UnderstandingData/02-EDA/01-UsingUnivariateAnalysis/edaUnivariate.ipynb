```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/train.csv')
df
```

### Categorical data

```python
# Countplot
sns.countplot(x='Survived', data=df)
df['Survived'].value_counts()
```

```python
sns.countplot(x='Pclass', data=df)
df['Pclass'].value_counts()
```

```python
sns.countplot(x='Sex', data=df)
df['Sex'].value_counts()
```

```python
sns.countplot(x='Embarked', data=df)
df['Embarked'].value_counts()
```

```python
# Piechart
plt.pie(df['Survived'].value_counts(), labels=['Not Survived', 'Survived'], autopct='%1.1f%%')
plt.show()
```

```python
plt.pie(df['Pclass'].value_counts(), labels=['Class 1', 'Class 2', 'Class 3'], autopct='%1.1f%%')
plt.show()
```

```python
plt.pie(df['Sex'].value_counts(), labels=['male', 'female'], autopct='%1.1f%%')
plt.show()
```

```python
plt.pie(df['Embarked'].value_counts(), labels=['S', 'C', 'Q'], autopct='%1.1f%%')
plt.show()
```

### Numerical Data

```python
# Histogram
plt.hist(df['Age'])
plt.show()
```

```python
# Displot
sns.displot(df['Age'], kde=True, stat='density')
```
