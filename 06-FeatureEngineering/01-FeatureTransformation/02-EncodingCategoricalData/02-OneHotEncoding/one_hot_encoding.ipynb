## Scikit learn One Hot Encoder

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 0:4], df.iloc[:, -1], test_size=0.2, random_state=42)
```

```python
X_train
```

```python
X_test
```

```python
ohe = OneHotEncoder()
```

```python
X_train_encoded = ohe.fit_transform(X_train[["fuel", "owner"]]).toarray()
X_train_encoded
```

```python
X_test_encoded = ohe.transform(X_test[["fuel", "owner"]]).toarray()
X_test_encoded
```

```python
X_train[["brand", "km_driven"]].values
```

```python
X_train_final = np.hstack([X_train_encoded, X_train[["brand", "km_driven"]].values])
X_test_final = np.hstack([X_test_encoded, X_test[["brand", "km_driven"]].values])
```
